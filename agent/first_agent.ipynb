```python
import nest_asyncio 
nest_asyncio.apply()

import os

from dotenv import load_dotenv
from pydantic_ai import Agent

load_dotenv()
```

```python
basic_agent = Agent("groq:openai/gpt-oss-120b")
```

```python
response = basic_agent.run_sync("Why the Earth is round in shape? Answer me in one sentence.")

response.output
```

```python
pirate_agent = Agent(
    "groq:openai/gpt-oss-120b",
    description="An ai agent with pirate personna",
    system_prompt=(
        "You are Captain Monkey D Luffy. Who is on a mission to find one peice "
        "You must answer every question like a king of pirates. "
        "Never drop your character."
    )
)

response = pirate_agent.run_sync("Explain how to get food on a ship.Give me answer in short sentence.")
print("Pirate agent Answer:",response.output)
```

```python
## Run sync method

summarize_agent = Agent(
    "groq:openai/gpt-oss-120b",
    system_prompt="Summerize the given text into exactly 3 words"
)

summary_response = summarize_agent.run_sync(
    "The quick brown fox jumps over the lazy dog in the green grass.The dog get shock and run away."

)

print("Sync output:--",summary_response.output)
```
