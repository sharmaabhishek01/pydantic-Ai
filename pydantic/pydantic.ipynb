# Enter pydantic 
Pydantic changes the rule.If you say something is an integer,pydantic guarantees it is an integer.If it isn't , Pydantic will stop the program and tell you exactly want went wrong.

```python
from pydantic import BaseModel

class Player(BaseModel):
    name : str
    score : int

player1 = Player(name="Abhishek sharma",score=85)
print("Player 1:",player1) 
```

```python
player2 = Player(name="Enzo kumar",score="55")
print("Player 2:",player2) 
```

```python
## This will validate the error

player3 = Player(name="Abhi",score="Sixty six")
```

# Making Robust Schema
Sometimes, checking if something is integer or string isn't enough.
What if we want an age to be greater then 30? What if want to be a password is strong?

```python
from pydantic import BaseModel , Field , field_validator

class UserSignUp(BaseModel):
    username : str = Field (min_length=3 , max_length=15)
    age : int = Field(ge=13)
    account_type : str = "free"

## first we make function and then we apply custom classes.
    @field_validator('username')    
    @classmethod
    def check_username_no_space(cls , value:str) -> str:
        if " " in value:
            raise ValueError("User name can not contain space!")

        return value    
```

```python
user1 = UserSignUp(
    username = "abhishek121",
    age = 21
)

print(user1)
```

```python
user2 = UserSignUp(
    username = "komal12",
    age = 10
)

print(user2)
```

```python
user3 = UserSignUp(
    username = "abhishek kumar",
    age = 27
)

print(user3)
```
